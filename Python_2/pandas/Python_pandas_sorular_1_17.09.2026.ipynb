## Veriyi inceleme (5. soru)

```python
import pandas as pd

# veriler
data = {
    "isim": ["Ali", "Ayşe", "Mehmet", "Fatma", "Can", "Zeynep", "Burak"],
    "yas": [24, 30, 22, 35, 28, 42, 26],
    "departman": ["Yazılım", "İK", "Yazılım", "Pazarlama", "Finans", "Yazılım", "İK"],
    "maas": [45000, 38000, 42000, 39000, 52000, 68000, 36000]
}

df = pd.DataFrame(data)

# 1. İlk 3 satırı görüntüleme (.head)
print("--- İlk 3 Satır (.head(3)) ---")
print(df.head(3))
print()

# 2. Son 2 satırı görüntüleme (.tail)
print("--- Son 2 Satır (.tail(2)) ---")
print(df.tail(2))
print()

# 3. DataFrame boyutunu öğrenme (.shape)
# Sonuç (satır_sayısı, sütun_sayısı) şeklinde bir tuple döner
print("--- Boyut (.shape) ---")
print(f"Toplam Satır: {df.shape[0]}, Toplam Sütun: {df.shape[1]}")
print(f"Shape çıktısı: {df.shape}")
print()

# 4. Sütunların veri tipleri ve boş olmayan (non-null) değer sayısı (.info)
print("--- Genel Bilgi ve Tipler (.info()) ---")
df.info()
print()

# 5. Sayısal sütunların istatistiksel özeti (.describe)
# Ortalama (mean), min, max, standart sapma (std) ve çeyreklikleri verir
print("--- Sayısal Özet (.describe()) ---")
print(df[["yas", "maas"]].describe())

# Sadece belirli bir sütunun min, max ve ortalamasını tekil almak istersek:
print("\n--- Tekil İstatistikler (Maaş) ---")
print(f"Ortalama Maaş : {df['maas'].mean():.2f}")
print(f"En Düşük Maaş : {df['maas'].min()}")
print(f"En Yüksek Maaş: {df['maas'].max()}")
```

## Sütun ve Satır Seçme (6. soru)

```python
import pandas as pd

# Örnek DataFrame
veri = {
    "isim": ["Ahmet", "Zeynep", "Mehmet", "Elif", "Can"],
    "yas": [22, 21, 23, 22, 25],
    "bolum": ["Yazılım", "Bilgisayar", "Yazılım", "Endüstri", "Makine"]
}

df = pd.DataFrame(veri)

# 1. Sadece "isim" sütununu seçme (Series döner)
print("--- 1. Sadece 'isim' Sütunu ---")
print(df["isim"])
print()

# 2. "isim" ve "yas" sütunlarını birlikte seçme (Liste içinde verilir, DataFrame döner)
print("--- 2. 'isim' ve 'yas' Sütunları ---")
print(df[["isim", "yas"]])
print()

# 3. .loc[] kullanarak ilk 3 satırı seçme (Etiket/İndeks bazlı: 0, 1 ve 2 dahil edilir)
print("--- 3. .loc[] ile İlk 3 Satır ---")
print(df.loc[0:2])
print()

# 4. .iloc[] kullanarak 2. satır ile 2. sütunun kesiştiği değeri bulma
# Python sıfır tabanlı (zero-indexed) olduğu için:
# 2. satır -> index 1
# 2. sütun -> index 1
deger = df.iloc[1, 1]
print("--- 4. .iloc[1, 1] Değeri ---")
print("Kesişim Değeri:", deger)
```

## Filtreleme (Koşullu Seçim) (7. soru)

```python
import pandas as pd

# Faz 1: Mock veri setinin hazırlanması
veri = {
    "isim": ["Ahmet", "Zeynep", "Mehmet", "Elif", "Can", "Burak", "Selin"],
    "yas": [22, 28, 19, 32, 45, 26, 23],
    "sehir": ["İstanbul", "Ankara", "İstanbul", "İzmir", "Bursa", "İstanbul", "Antalya"],
    "maas": [22000, 29000, 18000, 31000, 48000, 24500, 21000]
}

df = pd.DataFrame(veri)

# Faz 2: Koşullu filtreleme operasyonları

# 1. Yaş > 25 kontrolü
yas_filtre = df[df["yas"] > 25]
print("--- Yaşı 25'ten Büyükler ---")
print(yas_filtre)

# 2. Maaş aralığı kontrolü (20000 <= maas <= 30000)
maas_filtre = df[(df["maas"] >= 20000) & (df["maas"] <= 30000)]
print("\n--- Maaşı 20000 ile 30000 Arasında Olanlar ---")
print(maas_filtre)

# 3. Şehir bazlı filtreleme (İstanbul)
sehir_filtre = df[df["sehir"] == "İstanbul"]
print("\n--- İstanbul'da Yaşayanlar ---")
print(sehir_filtre)

# 4. Alternatif yaş sınırları kontrolü (yas < 20 VEYA yas > 40)
uc_yas_filtre = df[(df["yas"] < 20) | (df["yas"] > 40)]
print("\n--- Yaşı 20'den Küçük veya 40'tan Büyük Olanlar ---")
print(uc_yas_filtre)
```

## Yeni Sütun Ekleme ve Güncelleme (8. soru)

```python
import pandas as pd

# Faz 1: Mock veri setinin hazırlanması
veri = {
    "isim": ["Ahmet", "Zeynep", "Mehmet", "Elif", "Can"],
    "yas": [24, 16, 22, 17, 30],
    "maas": [35000, 15000, 42000, 18000, 55000]
}

df = pd.DataFrame(veri)

# Faz 2: Sütun ekleme ve manipülasyon adımları

# 1. Vektörize matematiksel işlemle doğum yılı sütununun eklenmesi
df["dogum_yili"] = 2026 - df["yas"]

# 2. Mevcut maaş sütununun in-place güncellenmesi (%10 zam)
df["maas"] = df["maas"] * 1.10

# 3. apply() ve lambda fonksiyonu ile koşullu kategorik sütun türetme
df["durum"] = df["yas"].apply(lambda y: "Yetişkin" if y >= 18 else "Çocuk")

# Faz 3: Nihai çıktının raporlanması
print(df)
```

## Sıralama (9. soru + bonus soru)

```python
import pandas as pd

# Faz 1: Sıralama operasyonları için veri seti
veri = {
    "isim": ["Ahmet", "Zeynep", "Mehmet", "Elif", "Can", "Burak"],
    "yas": [24, 21, 28, 21, 30, 24],
    "sehir": ["İstanbul", "Ankara", "İstanbul", "Ankara", "İzmir", "İstanbul"],
    "maas": [32000, 45000, 28000, 39000, 52000, 34000]
}

df = pd.DataFrame(veri)

# 1. Yaşa göre küçükten büyüğe sıralama (ascending=True varsayılandır)
df_yas_sirali = df.sort_values(by="yas")
print("--- Yaşa Göre Artan Sıralama ---")
print(df_yas_sirali)

# 2. Maaşa göre büyükten küçüğe sıralama
df_maas_sirali = df.sort_values(by="maas", ascending=False)
print("\n--- Maaşa Göre Azalan Sıralama ---")
print(df_maas_sirali)

# 3. Çoklu sütun sıralama (önce sehir A->Z, aynı şehirdekiler yaşa göre küçükten büyüğe)
df_coklu_sirali = df.sort_values(by=["sehir", "yas"], ascending=[True, True])
print("\n--- Önce Şehir, Sonra Yaş Sıralaması ---")
print(df_coklu_sirali)


# Faz 2: Bonus Mini Ödev
ogrenci_verisi = {
    "isim": ["Emir", "Yusuf", "Sena", "Kaan", "Buse", "Mert"],
    "yas": [23, 21, 22, 24, 20, 25],
    "sehir": ["İstanbul", "Ankara", "İzmir", "Bursa", "İstanbul", "Antalya"],
    "not1": [85, 60, 90, 45, 95, 70],
    "not2": [75, 65, 80, 55, 85, 60],
    "not3": [90, 70, 85, 60, 90, 65]
}

df_ogrenciler = pd.DataFrame(ogrenci_verisi)

# 1. Not ortalamasını vektörize olarak hesaplayıp yeni sütun açma
df_ogrenciler["ortalama"] = (df_ogrenciler["not1"] + df_ogrenciler["not2"] + df_ogrenciler["not3"]) / 3

# 2. Ortalaması 70 üstü olanları filtreleme
basarili_ogrenciler = df_ogrenciler[df_ogrenciler["ortalama"] > 70]

# 3. Filtrelenen veriyi yaşa göre küçükten büyüğe sıralama
sonuc = basarili_ogrenciler.sort_values(by="yas", ascending=True)

# Faz 3: Raporlama
print("\n--- Bonus Ödev: Ortalaması 70 Üstü Olanlar (Yaşa Göre Sıralı) ---")
print(sonuc[["isim", "yas", "sehir", "ortalama"]])
```
