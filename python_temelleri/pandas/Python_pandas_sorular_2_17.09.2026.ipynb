---## Bölüm 6 — Filtreleme ve Yeni Sütun**Soru 6.1:** Toplam tutarı 300 TL'nin üzerinde olan siparişleri filtreleyin.**Soru 6.2:** Ürün adedi 2'den fazla VE toplam tutar 200 TL'den fazla olan siparişleri bulun (`&` kullanın).**Soru 6.3:** `birim_fiyat` adında yeni bir sütun ekleyin; değeri `toplam_tutar / urun_adedi` olsun.**Soru 6.4:** `apply` ve `lambda` kullanarak kargo süresi 3 günden fazlaysa "Yavaş", değilse "Hızlı" yazan bir `kargo_durumu` sütunu oluşturun.

```python
# 6.1
filtre_300 = df[df["toplam_tutar"] > 300]
filtre_300


# 6.2
filtre_kombin = df[(df["urun_adedi"] > 2) & (df["toplam_tutar"] > 200)]
filtre_kombin

# 6.3
df["birim_fiyat"] = df["toplam_tutar"] / df["urun_adedi"]

# 6.4
df["kargo_durumu"] = df["kargo_suresi"].apply(lambda x: "Yavaş" if x > 3 else "Hızlı")

df
```

---## Bölüm 8 — Eksik Veri (NaN)**Soru 8.1:** `kargo_suresi` sütununda kaç eksik değer olduğunu bulun.**Soru 8.2:** Eksik kargo süresini sütunun **medyanı** ile doldurup `df_dolu` adlı yeni bir DataFrame oluşturun.**Soru 8.3:** Eksik değer içeren satırı silerek `df_temiz` adlı yeni bir DataFrame oluşturun ve satır sayısının değiştiğini `.shape` ile doğrulayın.

```python
# 8.1
eksik_sayisi = df["kargo_suresi"].isna().sum()
print("Eksik kargo süresi sayısı:", eksik_sayisi)

# 8.2
medyan_deger = df["kargo_suresi"].median()
df_dolu = df.copy()
df_dolu["kargo_suresi"] = df_dolu["kargo_suresi"].fillna(medyan_deger)

# 8.3
df_temiz = df.dropna().copy()
print("Orijinal shape:", df.shape, "Temiz shape:", df_temiz.shape)
```

---## Bölüm 9 — Merge (Tabloları Birleştirme)Aşağıdaki iki tabloyu kullanın:

```python
# 9.1# 9.2# 9.3

df_inner_join = pd.merge(musteriler, siparisler, on="musteri_id", how="inner")

df_left_join = pd.merge(musteriler, siparisler, on="musteri_id", how="left")

df_outer_join = pd.merge(musteriler, siparisler, on="musteri_id", how="outer")
```

---## Bölüm 11 — GroupByAşağıdaki veriyi kullanın:

```python
df_calisan = pd.DataFrame({    "sube": ["Kadıköy", "Beşiktaş", "Kadıköy", "Şişli", "Beşiktaş"],    "ad": ["Baran", "Ceylin", "Emre", "Fulya", "Gökhan"],    "satis": [12000, 18500, 9500, 21000, 16000]})
```

**Soru 11.1:** Şubeye göre **toplam** satışı bulun.**Soru 11.2:** Şubeye göre **ortalama** satışı bulun.**Soru 11.3:** `.agg(["sum","mean","count"])` kullanarak şube başına üç istatistiği aynı anda hesaplayın.

```python
# 11.1
print(df_calisan.groupby("sube")["satis"].sum())

# 11.2
print(df_calisan.groupby("sube")["satis"].mean())

# 11.3
print(df_calisan.groupby("sube")["satis"].agg(["sum", "mean", "count"]))
```

```python
# 12.1
pivot_1 = df_satis.pivot_table(index="sehir", columns="kategori", values="satis")

# 12.2: Kesişmeyen kombinasyonlara NaN yerine 0 atanır
pivot_2 = df_satis.pivot_table(
    index="sehir",
    columns="kategori",
    values="satis",
    aggfunc="sum",
    fill_value=0,
)
```

---## Bölüm 13 — Apply ile String İşlemleri**Soru 13.1:** `df_kullanici = pd.DataFrame({"kullanici_adi": ["baran23", "CEYLIN_K", "emre.y"]})` oluşturun.**Soru 13.2:** `apply(str.lower)` kullanarak tüm kullanıcı adlarını küçük harfe çeviren bir `kullanici_adi_kucuk` sütunu ekleyin.**Soru 13.3:** `apply` ve `lambda` ile kullanıcı adının kaç karakter olduğunu gösteren bir `karakter_sayisi` sütunu ekleyin.

```python
# 13.1
df_kullanici = pd.DataFrame(
    {"kullanici_adi": ["baran23", "CEYLIN_K", "emre.y"]}
)

# 13.2
df_kullanici["kullanici_adi_kucuk"] = df_kullanici["kullanici_adi"].apply(
    str.lower
)

# 13.3
df_kullanici["karakter_sayisi"] = df_kullanici["kullanici_adi"].apply(
    lambda x: len(x)
)
```

---## Bölüm 15 — Matplotlib ile Grafik ÇizmeAşağıdaki hücreyi çalıştırarak bir yıllık çeyreklik gelir verisi oluşturun:

```python
df_gelir = pd.DataFrame({    "ceyrek": ["Ç1", "Ç2", "Ç3", "Ç4"],    "gelir": [42000, 55000, 48000, 61000],    "gider": [30000, 38000, 33000, 40000]})
```

**Soru 15.1:** `gelir` değerlerini `kind="line"` ile, üzerinde nokta işaretleri (`marker="o"`) olacak şekilde çizgi grafikle çizin; başlık ve eksen etiketleri ekleyin.**Soru 15.2:** `gelir` değerlerini çeyreklere göre `kind="bar"` ile bar grafikle çizin (renk seçiminiz size ait).**Soru 15.3:** `gelir` sütununu çeyreklerin payını gösterecek şekilde `kind="pie"` ile pasta grafikle çizin (`autopct="%1.1f%%"` kullanın).**Soru 15.4:** `gelir` ve `gider` sütunlarını `kind="scatter"` ile scatter grafikte karşılaştırın (x ekseni gelir, y ekseni gider olsun) ve grafiğin ne anlattığını bir cümleyle yorumlayın.

```python
import matplotlib.pylab as plt
# 15.1
df_gelir.plot(x="ceyrek", y="gelir", kind="line", marker="o")
plt.title("Çeyreklik Gelir Dağılımı")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir (TL)")
plt.show()

# 15.2
df_gelir.plot(x="ceyrek", y="gelir", kind="bar", color="skyblue")
plt.title("Çeyreklik Gelir Bar Grafiği")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir (TL)")
plt.show()

# 15.3
df_gelir.set_index("ceyrek")["gelir"].plot(kind="pie", autopct="%1.1f%%")
plt.title("Çeyreklik Gelir Dağılım Payı")
plt.ylabel("")
plt.show()

# 15.4
df_gelir.plot(kind="scatter", x="gelir", y="gider")
plt.title("Gelir vs Gider Dağılımı")
plt.show()
# Yorum: Gelir ile gider arasında pozitif yönlü doğrusal bir ilişki bulunmaktadır; gelir arttıkça gider de düzenli olarak artmaktadır.
```

---## Bonus — Kapsamlı UygulamaBir online kursun satış verisini simüle edin:1. `kurs_adi`, `kategori` (örn. Yazılım, Tasarım, Pazarlama) ve `satis_adedi` sütunlarından oluşan, en az 8 satırlık bir DataFrame oluşturun.2. Kategoriye göre toplam satış adedini `groupby` ile bulun.3. En çok satan 3 kursu `.sort_values()` ile bulun.4. Kategori bazında toplam satışı gösteren bir bar grafik çizin.5. `apply` ile satış adedi 50 üzeri olanlara "Popüler", altındakilere "Standart" yazan yeni bir sütun ekleyin.

```python
# ==============================================================================
# Bonus — Kapsamlı Uygulama
# ==============================================================================
# 1. DataFrame oluşturma
df_kurs = pd.DataFrame(
    {
        "kurs_adi": [
            "Python Sıfırdan",
            "İleri Java",
            "UI/UX Tasarım",
            "Dijital Pazarlama",
            "FastAPI ile Backend",
            "Figma Masterclass",
            "SEO Eğitimi",
            "Spring Boot",
        ],
        "kategori": [
            "Yazılım",
            "Yazılım",
            "Tasarım",
            "Pazarlama",
            "Yazılım",
            "Tasarım",
            "Pazarlama",
            "Yazılım",
        ],
        "satis_adedi": [120, 45, 60, 30, 85, 40, 25, 95],
    }
)

# 2. Kategori bazında toplam satış
toplam_satis_kategori = df_kurs.groupby("kategori")["satis_adedi"].sum()
print("Kategori Bazında Satışlar:\n", toplam_satis_kategori)

# 3. En çok satan 3 kurs
en_cok_satan_3 = df_kurs.sort_values(by="satis_adedi", ascending=False).head(3)
print("\nEn Çok Satan 3 Kurs:\n", en_cok_satan_3)

# 4. Bar grafik
toplam_satis_kategori.plot(kind="bar", color="teal")
plt.title("Kategori Bazında Toplam Satış Adetleri")
plt.xlabel("Kategori")
plt.ylabel("Toplam Satış")
plt.show()

# 5. Koşullu sütun ekleme
df_kurs["durum"] = df_kurs["satis_adedi"].apply(
    lambda x: "Popüler" if x > 50 else "Standart"
)
print("\nNihai Kurs Veri Seti:\n", df_kurs)
```
