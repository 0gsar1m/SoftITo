```python
class Araba:
    def __init__(self, marka, model, hiz_kms=0):
        self.marka = marka
        self.model = model
        self.hiz_kms = hiz_kms

    def hizlandir(self, add):
        self.hiz_kms += add

    def bilgi_goster(self):
        print(f"{self.marka} {self.model} - Hız: {self.hiz_kms} km/s")


arac = Araba("VW", "Scirocco", 170)
arac.hizlandir(40)
arac.bilgi_goster()
```

```python
class Dikdortgen:
    def __init__(self, uzunluk, genislik):
        self.genislik = genislik;
        self.uzunluk = uzunluk;
        

    def alan(self):
        return self.uzunluk * self.genislik

    def cevre(self):
        return 2 * (self.uzunluk + self.genislik)


dikdortgen = Dikdortgen(10, 5)

print("Alan: ", dikdortgen.alan())
print("Çevre: ", dikdortgen.cevre())
```

```python
class Kitap:
    def __init__(self, ad, yazar, sayfa):
        self.ad = ad;
        self.sayfa = sayfa;
        self.yazar = yazar;

    def ozet(self):
        return f"{self.ad} - {self.yazar} ({self.sayfa} sayfa)"


kitap = Kitap("Dil Belası", "İmam Gazali rh.", 123)
print(kitap.ozet())
```
